```python
frutas = ["banana", "maçã", "laranja", "uva", "abacaxi"]
```

```python
for index, fruta in enumerate(frutas):
    print(f"{index}: {fruta}")
```

```python
numeros = [10, 20, 30, 40, 50]
```

```python
indice_valor = {index: x for index, x in enumerate(numeros)}
indice_par = {index: x for index, x in enumerate(numeros) if index % 2 == 0}
indice_impar = {index: x for index, x in enumerate(numeros) if index % 2 != 0}

print(f"Índice e valor: {indice_valor}")
print(f"Índice e valor (pares): {indice_par}")
print(f"Índice e valor (ímpares): {indice_impar}")
```
