```python
nomes = ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Heidi", "Ivan", "Judy"]
idades = [25, 30, 22, 28, 35, 27, 24, 29, 31, 26]
calcado = [38, 42, 40, 41, 39, 43, 37, 44, 36, 45]
```

```python
for nome, idade, calc in zip(nomes, idades, calcado):
    print(f"Nome: {nome}, Idade: {idade}, Calçado: {calc}")
```

```python
pessoas = list(zip(nomes, idades, calcado))
pessoas_index = enumerate(pessoas)
list(pessoas_index)
```
