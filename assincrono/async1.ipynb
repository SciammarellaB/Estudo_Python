```python
import asyncio
import aiohttp
```

```python
async def tarefa(nome, tempo):
    print(f'Tarefa {nome} iniciada, aguardando {tempo} segundos...')
    await asyncio.sleep(tempo)
    print(f'Tarefa {nome} concluída!')
```

```python
async def main():
    await asyncio.gather(
        tarefa('A', 2),
        tarefa('B', 3),
        tarefa('C', 1)
    )
```

```python
try:
    asyncio.get_running_loop()
except RuntimeError:
    asyncio.run(main())
```

```python
asyncio.ensure_future(main())
```

```python
async def pegar_usuario(session, usuario_id):
    async with session.get(
        f'https://jsonplaceholder.typicode.com/users/{usuario_id}'
    ) as response:
        return await response.json()
```

```python
async def main():
    async with aiohttp.ClientSession() as session:
            ids = [1, 2, 3, 4, 5]
            tarefas = [pegar_usuario(session, id) for id in ids]
            resultados = await asyncio.gather(*tarefas)
            for resultado in resultados:
                print(resultado)
```

```python
try:
    asyncio.get_running_loop()
except RuntimeError:
    asyncio.run(main())
```

```python
asyncio.ensure_future(main())
```
