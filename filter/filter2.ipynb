```python
def em_estoque(x):
    return x['estoque'] > 0
```

```python
produtos = [
    {"nome": "Produto A", "preco": 10.0, "estoque": 10},
    {"nome": "Produto B", "preco": 15.0, "estoque": 0},
    {"nome": "Produto C", "preco": 20.0, "estoque": 5},
    {"nome": "Produto D", "preco": 25.0, "estoque": 0},
    {"nome": "Produto E", "preco": 30.0, "estoque": 3}
]

produto_estoque = filter(em_estoque, produtos)
list(produto_estoque)
```
