```python
def soma(a: int, b: int) -> int:
    return a + b

def subtracao(a: int, b: int) -> int:
    return a - b

def multiplicacao(a: int, b: int) -> int:
    return a * b

def divisao(a: int, b: int) -> float:
    if b == 0:
        raise ValueError("Divisão por zero não é permitida.")
    return a / b

def media(notas: list[float]) -> float:
    return sum(notas) / len(notas)

class Estudante:
    def __init__(self, nome: str, idade: int, notas: list[float]):
        self.nome = nome
        self.idade = idade
        self.notas = notas

    def calcular_media(self) -> float:
        return media(self.notas)

    def __str__(self):
        return f"Estudante(nome = {self.nome}, idade = {self.idade}, media = {self.calcular_media():.2f})"
```

```python
soma = soma(10, 12)
subtracao = subtracao(10, 12)
multiplicacao = multiplicacao(10, 12)
divisao = divisao(10, 12)
notas = media ([10, 12, 14, 16, 18])

print(f"Soma: {soma}")
print(f"Subtração: {subtracao}")
print(f"Multiplicação: {multiplicacao}")
print(f"Divisão: {divisao}")
print(f"Média: {notas}")

print("----------------------------")

estudante1 = Estudante("João", 20, [5, 8, 9, 10, 3])
estudante2 = Estudante("Maria", 22, [7, 9, 10, 8, 6])
estudante3 = Estudante("Pedro", 19, [6, 7, 8, 9, 10])
estudante4 = Estudante("Ana", 21, [9, 10, 8, 7, 6])
estudante5 = Estudante("Lucas", 23, [10, 9, 8, 7, 6])

print(estudante1)
print(estudante2)
print(estudante3)
print(estudante4)
print(estudante5)
```
