```python
def numeros_impares(n):
    for i in range(1, n+1, 2):
        yield i
```

```python
for numero in numeros_impares(10):
    print(numero)
```

```python
def linhas_com_numeros_palavras(arquivo, numero):
    with open(arquivo, 'r', encoding='utf-8') as f:
        for linha in f:
            palavras = linha.split()
            if len(palavras) == numero:
                yield linha.strip()
```

```python
arquivo = 'texto.txt'
for linha in linhas_com_numeros_palavras(arquivo, 2):
    print(linha)
```
